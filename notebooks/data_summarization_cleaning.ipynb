```python
# Create a list of summaries for each column
unique_summary = []

for i in df.columns:
    unique_vals = df[i].dropna().unique()
    total_unique = len(unique_vals)
    
    if total_unique > 5:
        preview = f"{list(unique_vals[:5])} ... (+{total_unique - 5} more)"
    else:
        preview = list(unique_vals)
        
    unique_summary.append({
        'Column Name': i,
        'Total Unique Count': total_unique,
        'Unique Values Preview': preview
    })

df_unique_map = pd.DataFrame(unique_summary)
df_unique_map
```

We created a custom function to generate a summary DataFrame showing the unique counts and values for each column. This allows us to quickly audit our data for two critical checks: identifying impossible negative values in columns like children, and verifying that binary columns correctly contain only two distinct values.

ADR stands for Average Daily Rate. It is one of the most critical metrics in the hotel industry. It tells you the average revenue earned per occupied room on a given day.

customer_type:

• Transient: A walk-in or standard individual booking. They are not part of any group, event, or long-term contract.

• Transient-Party: An individual booking, but it is explicitly linked to another transient booking (e.g., a family booking two separate rooms under different names).

• Contract: Bookings tied to a fixed, recurring allocation contract (e.g., airline pilots/crew who have a standing room agreement every week).

• Group: A block booking made for an explicit group, such as a wedding party or a tour group.

```python
# 1. Select only the numeric columns
numeric_cols = df.select_dtypes(include=['int64', 'float64', 'Int64']).columns

# 2. Find which columns have values less than 0
negative_counts = (df[numeric_cols] < 0).sum()

# 3. Filter to display only columns that actually contain negatives
columns_with_negatives = negative_counts[negative_counts > 0]

if columns_with_negatives.empty:
    print("No negative values found in any numeric columns.")
else:
    print("Found negative values in the following columns:")
    print(columns_with_negatives)
```
